```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
np.random.seed(10)
iris = datasets.load_iris()
wine = datasets.load_wine()
```

```python
iris.keys()
```

```python
wine.keys()
```

```python
iris_data = iris.data
iris_target = iris.target
wine_data = wine.data
wine_target = wine.target
```

```python
iris_data_test, iris_data_train, iris_target_test, iris_target_train = train_test_split(iris_data, iris_target, test_size=0.2)
wine_data_test, wine_data_train, wine_target_test, wine_target_train = train_test_split(wine_data, wine_target, test_size=0.2)
```

```python
print(iris_data_test.shape)
print(iris_data_train.shape)
print(iris_target_test.shape)
print(iris_target_train.shape)
```

```python
print(wine_data_test.shape)
print(wine_data_train.shape)
print(wine_target_test.shape)
print(wine_target_train.shape)
```

```python
models = {
     "Logistic Regression": LogisticRegression(max_iter=200),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=42)
}
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
for name, model in models.items():
    model.fit(iris_data_train, iris_target_train)

    train_pred = model.predict(iris_data_train)
    test_pred = model.predict(iris_data_test)

    train_acc = accuracy_score(iris_target_train, train_pred)
    test_acc = accuracy_score(iris_target_test, test_pred)

    print(f"\n{name}")
    print(f"Accuracy train: {train_acc:.3f}")
    print(f"Accuracy test:  {test_acc:.3f}")
```

```python
for name, model in models.items():
    model.fit(wine_data_train, wine_target_train)

    train_pred = model.predict(wine_data_train)
    test_pred = model.predict(wine_data_test)

    train_acc = accuracy_score(wine_target_train, train_pred)
    test_acc = accuracy_score(wine_target_test, test_pred)

    print(f"\n{name}")
    print(f"Accuracy train: {train_acc:.3f}")
    print(f"Accuracy test:  {test_acc:.3f}")
```

```python
model = LogisticRegression(max_iter=200)
model.fit(iris_data_train, iris_target_train)

pred = model.predict(iris_data_test)

print(confusion_matrix(iris_target_test, pred))
print(classification_report(iris_target_test, pred))
```
